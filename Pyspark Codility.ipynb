Transforming column to remove commas in records

```python
from pyspark.sql.functions import col, regexp_replace

df_county = df_county.withColumn(
    "REGISTERED_VOTERS",
    regexp_replace(col("REGISTERED_VOTERS"), ",", "").cast("long")
)
```
